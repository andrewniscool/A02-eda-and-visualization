1. Show that $m(a + bX) = a+b \times m(X)$.
Using the original formula:
$$m(X) = \dfrac{1}{N} \sum_{i=1}^N x_i$$
we can simplify
$$m(a + bX) = \frac{1}{N}\sum_{i=1}^{N}(a + bx_i)$$

$= \frac{1}{N}\left(\sum_{i=1}^{N}a + \sum_{i=1}^{N}bx_i\right)$

$= \frac{1}{N}\left(Na + b\sum_{i=1}^{N}x_i\right)$

$= a + b\left(\frac{1}{N}\sum_{i=1}^{N}x_i\right)$

$= a + b \times m(X)$

---
2. Show that $\text{cov}(X,a+bY) = b \times \text{cov}(X,Y)$.
Using the original formula:
$$
\text{cov}(X,Y) = \frac{1}{N} \sum_{i=1}^N (x_i - m(X))(y_i - m(Y)).
$$
We can simplify:
$$\text{cov}(X,a+bY)=\frac{1}{N}\sum_{i=1}^{N}(x_i-m(X))((a+by_i)-(a+b\,m(Y)))$$

$= \text{cov}(X,a+bY)=\frac{1}{N}\sum_{i=1}^{N}(x_i-m(X))((by_i-b\,m(Y)))$

$= \text{cov}(X,a+bY)=\frac{1}{N}\sum_{i=1}^{N}(x_i-m(X))(b(y_i-m(Y)))$

$= \text{cov}(X,a+bY)=b * \frac{1}{N}\sum_{i=1}^{N}(x_i-m(X))(y_i-m(Y))$

$= \text{cov}(X,a+bY)=b * \text{cov}(X,Y)$

---
3. Show that $\text{cov}(a+bX,a+bX) = b^2 \text{cov}(X,X) $, and in particular that $\text{cov}(X,X) = s^2 $. Using the original formulas
$$
\text{cov}(X,Y) = \dfrac{1}{N} \sum_{i=1}^N (x_i - m(X))(y_i - m(Y)).
$$
we can simplify:
$$\text{cov}(a+bX,a+bX)$$

$= \frac{1}{N}\sum_{i=1}^{N}((a+bx_i)-m(a+bX))((a+bx_i)-m(a+bX))$

$= \frac{1}{N}\sum_{i=1}^{N}((a+bx_i) - (a + m(bX)))((a+bx_i) - (a + m(bX)))$

$= \frac{1}{N}\sum_{i=1}^{N}(bx_i-m(bX))(bx_i-m(bX))$

$= \frac{1}{N}\sum_{i=1}^{N}(bx_i-(b * m(X)))(bx_i-(b * m(X)))$

$= \frac{1}{N}\sum_{i=1}^{N}(bx_i-(b * m(X))) ^ 2$

$= \frac{1}{N}\sum_{i=1}^{N}(b * (x_i-m(X))) ^ 2$

$= \frac{1}{N}\sum_{i=1}^{N}(b ^ 2 * (x_i-m(X))^ 2)$

Since:

$$\text{cov}(X, X) = (x_i-m(X))^ 2$$

Then:

$$b ^ 2 * \frac{1}{N}\sum_{i=1}^{N}(x_i-m(X)) ^ 2$$
$= b ^ 2 * \text{cov}(X, X)$

So:

$$\text{cov}(a+bX,a+bX) = b^2 \text{cov}(X,X)$$

then, using this formula:
$$
s^2 = \dfrac{1}{N} \sum_{i=1}^N (x_i - m(X))^2.
$$
we can simplify the previous formula into:

$$\frac{1}{N}\sum_{i=1}^{N}(b * (x_i-m(X))) ^ 2$$

$$= b ^ 2 * \frac{1}{N}\sum_{i=1}^{N}(x_i-m(X)) ^ 2$$

$$= b ^ 2 * s ^ 2$$

so:
$\text{cov}(X,X) = s^2 $

4. Create a kernel density plot of the log of `Foreign Gift Amount`, and then a kernel density plot of the log of `Foreign Gift Amount` conditional on gift type. Do you notice any patterns?
---
Yes, Contract gifts seem to take up the most density by far, followed by Monetary Gifts, and the Real estate gifts. Real Estate gifts take little to almost none of the density, and this data aligns well with the proportions found in question 2.3.

```python
foreign_gifts_df["Log Foreign Gift Amount"] = np.log(foreign_gifts_df["Foreign Gift Amount"])

sns.kdeplot(data=foreign_gifts_df, x="Log Foreign Gift Amount")

plt.xlabel("Log Foreign Gift Amount")
plt.title("Density of Log Foreign Gift Amount")
plt.show()
```

```python
sns.kdeplot(data=foreign_gifts_df, x="Log Foreign Gift Amount", hue="Gift Type")

plt.xlabel("Log Foreign Gift Amount")
plt.title("Density of Log Foreign Gift Amount by Gift Type")
plt.show()
```
